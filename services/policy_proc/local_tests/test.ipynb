```python
import boto3
import json

s3vectors = boto3.client("s3vectors", region_name="ap-southeast-2")

response = s3vectors.list_vectors(
    vectorBucketName="sena-s3-vectors-bucket",
    indexName="my-rag-index",
    returnData=False,
    returnMetadata=True,
    maxResults=5
)

for v in response.get("vectors", []):
    meta = v.get("metadata", {})
    bedrock_meta = meta.get("AMAZON_BEDROCK_METADATA", "")
    if bedrock_meta:
        try:
            parsed = json.loads(bedrock_meta)
            source = parsed.get("source", {}).get("sourceLocation", "")
            print(f"Source: {source}")
        except:
            pass
    print(f"Keys: {list(meta.keys())}")
    print()
```

```python
import boto3
import json

s3vectors = boto3.client("s3vectors", region_name="ap-southeast-2")

response = s3vectors.list_vectors(
    vectorBucketName="sena-s3-vectors-bucket",
    indexName="my-rag-index",
    returnData=False,
    returnMetadata=True,
    maxResults=20
)

for v in response.get("vectors", []):
    meta = v.get("metadata", {})
    org_id = meta.get("org_id", "NOT FOUND")
    bedrock_meta = meta.get("AMAZON_BEDROCK_METADATA", "")
    try:
        parsed = json.loads(bedrock_meta)
        source = parsed.get("source", {}).get("sourceLocation", "").split("/")[-1]
    except:
        source = "unknown"
    print(f"org_id: {org_id} | file: {source}")
```
